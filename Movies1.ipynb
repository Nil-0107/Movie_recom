```python
import pandas as pd
import ast

def robust_convert(x):
    # If it's already a list, just extract names
    if isinstance(x, list):
        return " ".join([i['name'] for i in x if 'name' in i])

    # If it's a string, try to parse it
    if isinstance(x, str):
        try:
            data = ast.literal_eval(x)
            if isinstance(data, list):
                return " ".join([i['name'] for i in data if 'name' in i])
        except Exception as e:
            # Uncomment the line below to see why it's failing
            # print(f"Failed to parse: {x} | Error: {e}")
            return "unknown"

    return ""

df['genres'] = df['genres'].apply(robust_convert)

# Check the results
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_word=set(stopwords.words('english'))
lemmatizer=WordNetLemmatizer()
```

```python
import regex

def preprocess_text(text):
    #converting into lower case
    text=str(text).lower()
    #removing the punctuations
    text = regex.sub(r"\p{P}", "", text)

    words=text.split()
    #remove stopwords
    words=[word for word in words if word not in stop_word]

    #lemmatize
    words=[lemmatizer.lemmatize(word) for word in words]

    return " ".join(words)

df['tags']=df['tags'].apply(preprocess_text)
```

```python
df=df.reset_index(drop=True)
```

```python
indices=pd.Series(df.index,index= df['title']).drop_duplicates()
```

```python
indices
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(max_features=50000,ngram_range=(1,2) ,stop_words='english')
```

```python
tfidf_matrix= tfidf.fit_transform(df['tags'])
```

```python
tfidf_matrix
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend(title, n = 10):
  if title not in indices:
    return ['Movie not found']

  idx = indices[title]
  sim_score = cosine_similarity(tfidf_matrix[idx], tfidf_matrix).flatten()
  similar_idx = sim_score.argsort()[::-1][1:n+1]
  return df['title'].iloc[similar_idx]
```

```python
recommend('Scent Of a Movie',n=5)
```
